```python
import numpy as np

matrix = np.array([
    [5, 8, 2, 7],
    [1, 9, 4, 3],
    [6, 0, 11, 10],
    [12, 13, 14, 15]
])

print("The second row:", matrix[1, :])

print("The third column:", matrix[:, 2])

sub_table = matrix[1:3, 1:3]
print("Sub-table (R2-R3, C2-C3):\n", sub_table)
```

```python
import numpy as np

scores = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [88, 92, 79],
    [55, 65, 72]
])

scores[:, 1] += 5
print("Scores after increasing Physics marks:\n", scores)

scores[0, :] = [100, 100, 100]
print("\nScores after replacing first student's marks:\n", scores)

scores[:, 2] = 0
print("\nScores after setting CS column to zero:\n", scores)
```

```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
v = np.array([2, 4])

print("Element-wise multiplication (A * B):\n", A * B)

print("\nMatrix multiplication (A @ B):\n", A @ B)

print("\nA * v using broadcasting:\n", A * v)
```

```python
import numpy as np

np.random.seed(42)
random_matrix = np.random.randint(-5, 6, size=(4, 4))
print("Random 4x4 Matrix:\n", random_matrix)

unique_vals, counts = np.unique(random_matrix, return_counts=True)
print("\nUnique values:", unique_vals)
print("Counts of each unique value:", counts)

print("\nUnique rows of random matrix:\n", np.unique(random_matrix, axis=0))

a = np.array([[1, 0, 0], [1, 0, 0], [2, 3, 4]])
unique_rows = np.unique(a, axis=0)
print("\nUnique rows of array 'a':\n", unique_rows)
```
