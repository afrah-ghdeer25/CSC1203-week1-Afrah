```python
import numpy as np

A = np.arange(1, 21).reshape(4, 5)

print("Array A:\n", A)
print(f"Data Type: {A.dtype}")
print(f"Dimensions: {A.ndim}")
print(f"Shape: {A.shape}")
print(f"Total Elements: {A.size}")
print(f"Item Size: {A.itemsize}")

B = A.reshape(5, 4)
print("\nArray B (reshaped):\n", B)
```

```python
import numpy as np

a = np.array([[0., 0., 0.], [10., 10., 10.], [20., 20., 20.], [30., 30., 30.]])
b = np.array([0., 1., 2.])

c = a + b
print("Array c (a + b):\n", c)

mask = (c >= 22)
print("\nBoolean mask (values >= 22):\n", mask)

count = np.sum(mask)
print("\nHow many values in c are >= 22? ", count)
```

```python
import numpy as np

grades = np.array([[87, 96, 70], [100, 87, 90], [94, 77, 90], [100, 81, 82]])

print("--- Overall Statistics ---")
print(f"Sum: {grades.sum()}")
print(f"Min: {grades.min()}, Max: {grades.max()}")
print(f"Mean: {grades.mean():.2f}")
print(f"Std: {grades.std():.2f}, Var: {grades.var():.2f}")

print("\nMean per column (axis=0):", grades.mean(axis=0))
print("Mean per row (axis=1):", grades.mean(axis=1))
print("Max per row (axis=1):", grades.max(axis=1))
```

```python
import numpy as np

arr4 = np.array([[2, 4, 6], [1, 3, 5], [7, 9, 11]])

print("The Array:\n", arr4)
print("Sum of each row:", arr4.sum(axis=1))
print("Sum of each column:", arr4.sum(axis=0))
print("Overall mean of the matrix:", arr4.mean())
```

```python
import numpy as np

A = np.array([[10, 20, 30], [40, 50, 60]])
B = np.array([1, 2, 3])

print("Add B to A:\n", A + B)

print("\nMultiply A by 2:\n", A * 2)

print("\nSubtract 5 from all elements of A:\n", A - 5)
```

```python
import numpy as np

grades = np.array([[70, 85, 90], [88, 76, 95], [60, 80, 72]])

flat_max_idx = np.argmax(grades)
full_idx = np.unravel_index(flat_max_idx, grades.shape)
print("Index of the largest value (whole array):", full_idx)

print("Index of highest grade in each row:", np.argmax(grades, axis=1))

print("Index of lowest grade in each column:", np.argmin(grades, axis=0))
```
